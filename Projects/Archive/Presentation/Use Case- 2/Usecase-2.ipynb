## PDF to Excel 

```python
import tabula
import pandas as pd

# Path to the PDF file
pdf_file_path = r"C:/Users/chandresh.kanyal/Downloads/morningstarreport20241203092431.PDF"
output_excel_path = r"C:/Users/chandresh.kanyal/Downloads/extracted_tables_tabula_cleaned.xlsx"

# Extract all tables from the PDF
tables = tabula.read_pdf(pdf_file_path, pages='all', multiple_tables=True)

# Check if tables are found
if tables:
    # Create a Pandas ExcelWriter object to save all tables to an Excel file
    with pd.ExcelWriter(output_excel_path) as writer:
        # Iterate through the tables and save each one to a separate sheet
        for i, table in enumerate(tables):
            # Remove "Unnamed" columns (if any)
            table = table.loc[:, ~table.columns.str.contains('^Unnamed')]

            # Optionally, drop columns that are entirely empty (if needed)
            table = table.dropna(axis=1, how='all')
            
            # Dynamically create sheet names like Sheet1, Sheet2, etc.
            sheet_name = f"Sheet{i+1}"
            table.to_excel(writer, sheet_name=sheet_name, index=False)
    
    print(f"Tables extracted and saved to {output_excel_path}")
else:
    print("No tables were found in the PDF.")
```
