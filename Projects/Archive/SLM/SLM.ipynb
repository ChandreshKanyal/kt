```python
from transformers import pipeline

chatbot = pipeline("text2text-generation", model="google/flan-t5-small")
```

```python
def generate_response(message, history):
    response = chatbot(message, max_length=128, do_sample=False)[0]["generated_text"]
    return response.strip()
```

```python
generate_response("who is the prime minister of india?")
```

```python
import gradio as gr

gr.ChatInterface(
    fn=generate_response,
    title="FLAN-T5 Chatbot",
    description="Ask any question or give an instruction. Powered by google/flan-t5-small."
).launch(share=True)
```
